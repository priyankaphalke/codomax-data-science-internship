```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Display settings
pd.set_option("display.max_columns", None)

# Visualization style
sns.set_style("whitegrid")

print("Libraries imported successfully!")
```

```python
df = pd.read_csv("/SampleSuperstore.csv")
print("Dataset loaded successfully.")
```

```python
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
print("Columns in the dataset:")

for column in df.columns:
    print(column)
```

```python
df.info()
```

```python
df.describe()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
missing_percentage = (df.isnull().sum() / len(df)) * 100

print("Missing value percentage:")
print(missing_percentage)
```

```python
duplicates = df.duplicated().sum()

print("Number of duplicate rows:", duplicates)
```

```python
df = df.drop_duplicates()

print("Dataset shape after removing duplicates:", df.shape)
```

```python
print(df.dtypes)
```

```python
for column in df.select_dtypes(include="object").columns:
    print("\n", column)
    print(df[column].unique()[:10])
```

```python
numerical_columns = df.select_dtypes(include=np.number).columns

print("Numerical columns:")
print(list(numerical_columns))
```

```python
total_sales = df["Sales"].sum()

print("Total Sales:", total_sales)
```

```python
average_sales = df["Sales"].mean()

print("Average Sales:", average_sales)
```

```python
maximum_sale = df["Sales"].max()

print("Highest Sale:", maximum_sale)
```

```python
minimum_sale = df["Sales"].min()

print("Lowest Sale:", minimum_sale)
```

```python
total_quantity = df["Quantity"].sum()

print("Total Quantity Sold:", total_quantity)
```

```python
total_profit = df["Profit"].sum()

print("Total Profit:", total_profit)
```

```python
average_profit = df["Profit"].mean()

print("Average Profit:", average_profit)
```

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
)

print(category_sales)
```

```python
category_profit = df.groupby("Category")["Profit"].sum().sort_values(
    ascending=False
)

print(category_profit)
```

```python
category_quantity = df.groupby("Category")["Quantity"].sum().sort_values(
    ascending=False
)

print(category_quantity)
```

```python
plt.figure(figsize=(8, 5))

category_sales.plot(kind="bar")

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
plt.figure(figsize=(8, 5))

category_profit.plot(kind="bar")

plt.title("Total Profit by Category")
plt.xlabel("Category")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

```python
category_counts = df["Category"].value_counts()

plt.figure(figsize=(7, 7))

plt.pie(
    category_counts,
    labels=category_counts.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Product Category Distribution")

plt.show()
```

```python
subcategory_sales = df.groupby("Sub_Category")["Sales"].sum().sort_values(
    ascending=False
)

print(subcategory_sales)
```

```python
plt.figure(figsize=(10, 6))

subcategory_sales.plot(kind="bar")

plt.title("Sales by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Sales")

plt.xticks(rotation=45)

plt.show()
```

```python
subcategory_profit = df.groupby("Sub_Category")["Profit"].sum().sort_values(
    ascending=False
)

print(subcategory_profit)
```

```python
plt.figure(figsize=(10, 6))

subcategory_profit.plot(kind="bar")

plt.title("Profit by Sub-Category")
plt.xlabel("Sub-Category")
plt.ylabel("Profit")

plt.xticks(rotation=45)

plt.show()
```

```python
region_sales = df.groupby("Region")["Sales"].sum().sort_values(
    ascending=False
)

print(region_sales)
```

```python
plt.figure(figsize=(8, 5))

region_sales.plot(kind="bar")

plt.title("Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.xticks(rotation=0)

plt.show()
```

```python
region_profit = df.groupby("Region")["Profit"].sum().sort_values(
    ascending=False
)

print(region_profit)
```

```python
plt.figure(figsize=(8, 5))

region_profit.plot(kind="bar")

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Profit")

plt.xticks(rotation=0)

plt.show()
```

```python
top_cities = df.groupby("City")["Sales"].sum().sort_values(
    ascending=False
).head(10)

print(top_cities)
```

```python
plt.figure(figsize=(10, 6))

top_cities.sort_values().plot(kind="barh")

plt.title("Top 10 Cities by Sales")
plt.xlabel("Sales")
plt.ylabel("City")

plt.show()
```

```python
top_products = df.groupby("Category")["Sales"].sum().sort_values(
    ascending=False
).head(10)

print(top_products)
```

```python
plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Sales")
plt.xlabel("Sales")
plt.ylabel("Product")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(df["Sales"], bins=30)

plt.title("Distribution of Sales")
plt.xlabel("Sales")
plt.ylabel("Frequency")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit"
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")

plt.show()
```

```python
numerical_df = df.select_dtypes(include=np.number)

correlation = numerical_df.corr()

correlation
```

```python
segment_sales = df.groupby("Segment")["Sales"].sum().sort_values(ascending=False)

print("Sales by Customer Segment:")
print(segment_sales)
```

```python
segment_profit = df.groupby("Segment")["Profit"].sum().sort_values(ascending=False)

print("Profit by Customer Segment:")
print(segment_profit)
```

```python
plt.figure(figsize=(8, 5))

segment_sales.plot(kind="bar")

plt.title("Sales by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Total Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
print("Average Discount:", df["Discount"].mean())
print("Maximum Discount:", df["Discount"].max())
print("Minimum Discount:", df["Discount"].min())
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x="Discount",
    y="Profit"
)

plt.title("Discount vs Profit")
plt.xlabel("Discount")
plt.ylabel("Profit")

plt.show()
```

```python
region_category_sales = pd.pivot_table(
    df,
    values="Sales",
    index="Region",
    columns="Category",
    aggfunc="sum"
)

print(region_category_sales)
```

```python
region_category_sales.plot(
    kind="bar",
    figsize=(10, 6)
)

plt.title("Sales by Region and Category")
plt.xlabel("Region")
plt.ylabel("Sales")

plt.xticks(rotation=0)
plt.show()
```

```python
loss_making = df[df["Profit"] < 0]

print("Number of loss-making records:", len(loss_making))
print("Total Loss:", loss_making["Profit"].sum())
```

```python
profitable_products = df.groupby("Category")["Profit"].sum().sort_values(
    ascending=False
).head(10)

print(profitable_products)
```

```python
plt.figure(figsize=(10, 6))

profitable_products.sort_values().plot(kind="barh")

plt.title("Top 10 Most Profitable Products")
plt.xlabel("Profit")
plt.ylabel("Product")

plt.show()
```

```python
loss_products = df.groupby("Category")["Profit"].sum().sort_values(
    ascending=True
).head(10)

print(loss_products)
```

```python
plt.figure(figsize=(10, 6))

loss_products.sort_values().plot(kind="barh")

plt.title("Top 10 Loss-Making Products")
plt.xlabel("Profit")
plt.ylabel("Product")

plt.show()
```

```python
print("========== FINAL SUMMARY ==========")

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Total Quantity Sold:", df["Quantity"].sum())
print("Total Profit:", df["Profit"].sum())
print("Average Profit:", df["Profit"].mean())
print("Average Discount:", df["Discount"].mean())
print("Number of Records:", len(df))
```

```python
best_category = category_sales.idxmax()
best_region = region_sales.idxmax()
best_subcategory = subcategory_sales.idxmax()
best_segment = segment_sales.idxmax()

print("Best Performing Category:", best_category)
print("Best Performing Region:", best_region)
print("Best Performing Sub-Category:", best_subcategory)
print("Best Performing Customer Segment:", best_segment)
```
