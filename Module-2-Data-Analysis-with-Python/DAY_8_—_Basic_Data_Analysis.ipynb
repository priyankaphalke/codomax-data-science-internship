```python
import numpy as np
import pandas as pd
```

```python
data = {
    "Name": [
        "Aarav", "Priya", "Rahul", "Sneha", "Aditya",
        "Ananya", "Rohan", "Neha", "Vikram", "Isha"
    ],
    "Age": [20, 21, 19, 20, 22, 21, 20, 19, 22, 21],
    "Gender": [
        "Male", "Female", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Study_Hours": [3, 5, 2, 6, 4, 7, 3, 5, 2, 6],
    "Attendance": [85, 92, 78, 95, 88, 96, 82, 90, 75, 93],
    "Marks": [72, 88, 65, 91, 79, 95, 70, 86, 62, 90]
}

df = pd.DataFrame(data)

df
```

```python
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

```python
highest_marks = df["Marks"].max()

print("Highest Marks:", highest_marks)
```

```python
lowest_marks = df["Marks"].min()

print("Lowest Marks:", lowest_marks)
```

```python
average_study_hours = df["Study_Hours"].mean()

print("Average Study Hours:", average_study_hours)
```

```python
average_attendance = df["Attendance"].mean()

print("Average Attendance:", average_attendance)
```

```python
top_students = df.sort_values(
    by="Marks",
    ascending=False
)

top_students[["Name", "Marks"]].head()
```

```python
high_performers = df[df["Marks"] > 80]

high_performers[["Name", "Marks"]]
```

```python
gender_performance = df.groupby("Gender")["Marks"].mean()

print(gender_performance)
```

```python
study_analysis = df[["Study_Hours", "Marks"]]

study_analysis
```

```python
correlation = df[["Study_Hours", "Attendance", "Marks"]].corr()

correlation
```

```python
# # Key Insights

# Based on the analysis:

# 1. The average marks provide an overall measure of student performance.

# 2. Students with higher study hours generally show better academic
#    performance in this sample.

# 3. Attendance can be compared with marks to understand whether attendance
#    is associated with academic performance.

# 4. The highest and lowest performing students can be identified using
#    sorting and filtering.

# 5. Grouping the data by gender allows comparison of average performance
#    between groups.

# 6. Correlation analysis helps identify relationships between study hours,
#    attendance, and marks.
```
