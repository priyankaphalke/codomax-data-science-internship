```python
import numpy as np
import pandas as pd
```

```python
numbers = np.array([10,20,30,40,50])

print("Array", numbers)
print("Data Type: ", numbers.dtype)
print("Shape: ", numbers.shape)
```

```python
numbers = np.array([10, 20, 30, 40, 50])

print("Sum:", np.sum(numbers))
print("Mean:", np.mean(numbers))
print("Maximum:", np.max(numbers))
print("Minimum:", np.min(numbers))
```

```python
marks = np.array([
    [80, 85, 90],
    [75, 88, 92],
    [90, 95, 89]
])

print(marks)
print("Shape:", marks.shape)
```

```python
marks = pd.Series([50, 89, 78, 99, 67])
print(marks)
```

```python
data = {
    "Name": ["Ali", "Vicky", "sweety", "kartiki", "Shrisha"],
    "Age": [20, 34, 22, 21, 23],
    "Marks": [80, 99, 78, 89, 87]

    }

df = pd.DataFrame(data)
df
```

```python
print("First rows:")
print(df.head())

print("\nColumns:")
print(df.columns)

print("\nShape:")
print(df.shape)
```

```python
print(df.describe())
```
