```python
#Arithmetic Operators

c = 20
d = 10
print("Addition is: ", c+d)
print("Subtraction is: ", c-d)
print("Multiplication is : ", c*d)
print("Divison : ", c/d)
print("floor division is: ", c//d)
print("Modulus: ", c%d)
print("Power: ", c ** d)
```

```python
#input Function

name = input("Enter your Name: ")
print("Hello", name)

age = input("Enter your age: ")
print("ur age: ", int(age))

print(f"so you are {name} and your age is {age}")
print(f"you are the BEST {name}")
```

```python
#conditional statements

marks = 89

if marks >=90:
  grade = "A+"
elif marks >=80:
  grade = "A"
elif marks >=75:
  grade = "B"
elif marks >= 60:
  grade = "C"
else:
  grade = "D"

print("Grade : ", grade)
```

```python
#Mini Project

number = int(input("Enter a number: "))

if (number %2 == 0):
  print("Even Number")
else:
  print("odd number")
```
