```python
# ## What is Data Science?

# Data Science is the field of extracting useful insights and knowledge
# from data using programming, statistics, mathematics, and machine learning.

# A typical Data Science workflow includes:

# 1. Problem Definition
# 2. Data Collection
# 3. Data Cleaning
# 4. Exploratory Data Analysis
# 5. Feature Engineering
# 6. Model Building
# 7. Model Evaluation
# 8. Deployment
# 9. Monitoring
```

```python
# ## Types of Data

# ### Structured Data
# Data organized in rows and columns, such as databases and spreadsheets.

# ### Unstructured Data
# Data such as images, videos, audio, and text.

# ### Semi-structured Data
# Data that contains some organizational structure, such as JSON and XML.
```

```python
# ## Real-World Applications of Data Science

# - Healthcare – disease prediction and patient analysis
# - Finance – fraud detection and risk analysis
# - E-commerce – recommendation systems
# - Marketing – customer segmentation
# - Transportation – route optimization
# - Social Media – content recommendation
# - Education – student performance analysis
```

```python
import pandas as pd
```

```python
data ={
    "Name": ["Alice", "Bobby", "tom", "jeeyan", "Doraemon"],
    "Age": [24, 20, 19, 23, 20],
    "Marks": [90, 89, 99, 97, 91]

}
df = pd.DataFrame(data)
df
```

```python
print("First 5 names: ")
df.head()

```

```python
print("Data information: ")
df.info()
```

```python
print("Statistics:")
df.describe()
```

```python
average_marks = df["Marks"].mean()

print("Average Marks:", average_marks)
```

```python
#calculate the highest marks
highest_marks = df["Marks"].max()

print("Highest Marks: ", highest_marks)
```

```python
#Student with the highest marks

top_student = df.loc[df["Marks"].idxmax()]

print(top_student)
```

```python
#Simple Data Analysis
print("Average marks: ", df["Marks"].mean())
print("Average Age: ", df["Age"].mean())
print("Highest Marks: ", df["Marks"].max())
print("Lowest Marks: ", df["Marks"].min())
print("Number of students: ", len(df))
```
