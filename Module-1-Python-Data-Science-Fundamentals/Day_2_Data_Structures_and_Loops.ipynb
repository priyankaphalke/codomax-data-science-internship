```python
#Lists

fruits = ["apple", "banana", "cherry", "strawberry"]
print(fruits)
print(type(fruits))

print("The first fruit is : ",fruits[0])

fruits.append("Grapes")
print(fruits)

fruits.remove(fruits[2])
print(fruits)
```

```python
#list operations

numbers = [10, 20, 30, 40, 50]
print("Length: ", len(numbers))
print("Maximum: ", max(numbers))
print("Minimum: ", min(numbers))
print("sum is: ", sum(numbers))
```

```python
#Tuples
#combination of different types of datatypes in the tuple datatype

student = ("Priyanka", 20, "Data scientist", 40, 23)
print(student)
print(student[0])
print((student[3]))
```

```python
#tuple operations
tup1 = (1, 2)
tup2 = (3, 4)
print(tup1[-1])
print(tup1 + tup2)
print(tup1 in tup2)
print(tup1 not in tup2)
result = tup1 * 4
print(result)
print(len)
```

```python
#more tuple operations
num = ( 0, 1,2, 3,8, 5,5)
print(num[1:4])
print(num[0:4])
print(num.index(8))
print(num.count(5))
print(num[::-1])
print(num[::4])
print(num[:2])
```

```python
numiing = [ 10, 39, 22,32,432]
print(len(numiing))
print(sum(numiing))
print(max(numiing))
print(min(numiing))
print(sorted(numiing))
print(numiing[::-1])

del numiing[0]
print(numiing)
```

```python
#sets
#sets automatically removes the duplicate values

numbs = {1,2,3,2,3,4,5,5,6}
print(numbs)

names = {"rishi", "kartiki","surbhi", "rishi"}
print(names)

combine = {"rashi", 1, 1,2, 4,"rashi"}
print(combine)
```

```python
#Dictionaries

student = {
    "name": "Priyanka",
    "age": 20,
    "course": "Artificial Intelligence",
    "height": 5.5,
    "score": 99
}
print(student)
print("Score: ", student["score"])
print("Name: ", student["name"])
print(student.update(name="riya"))

print(student)
```

```python
#for loop in python

for i in student:
  print(student["name"])
  continue
  print(student["score"])
```

```python
for i in range(1, 11):
  if(i %2 == 0):
    print(i)
    continue
    print(i)
```

```python
#loop through the list

for fruit in fruits :
  print(fruit)
```

```python
number = 1
while number <= 4:
  print(number)
  number +=1
```

```python
#multiplication table usin the the for loop

print("Print the table using the for loop")
numbe = int(input("Enter the number: "))
for i in range(1, 11):
  print(numbe, "x", i , "=", numbe * i)
```

```python
digit = [10, 20, 34,2, 422, 4321]

for nun in digit:
  if (nun % 2 == 0):
    print(nun)

  else:
    continue
```
