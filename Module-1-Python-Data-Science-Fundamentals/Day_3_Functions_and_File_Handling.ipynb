```python
#FUNCTIONS IN THE PYTHON
def greet():
  print("Welcome")

greet()
```

```python
#function with the string parameter
def greet(name):
  print("Hello,", name)

greet("Priyanka")
```

```python
def greet(name, age):
  print("Hello,", name, "You just turned", age, "Congraulations")

greet("Priyanka", 20)
```

```python
#create a file

with open("student.txt", "w") as file:
  file.write("Name: Priyanka\n")
  file.write("Age: 20\n")
  file.write("Score: 99\n")

print("file created successfully.")
```

```python
#read the file
with open("student.txt", "r") as file:
  content = file.read()

print(content)
```

```python
#Append to file
with open("student.txt", "a") as file:
  file.write("This is the information related to the student Priyanka Phalke.\n")

print("Data added successfully.")
```

```python
#Read line by line
with open("student.txt", "r") as file:
  for line in file:
    print(line.strip())
```

```python
#mini Project
import builtins
def calculate_result(marks):
  average = builtins.sum(marks) / len(marks)

  if average >=90:
    grade = "A+"
  elif average >= 80:
    grade = "A"
  elif average >=70:
    grade = "B+"
  elif average >=60:
    grade = "B"
  elif average >=50:
    grade = "C"
  else:
    print("Fail")

  return average, grade


marks = [80, 88, 90, 99, 89, 90]

average, grade = calculate_result(marks)

print("Average marks: ", average)
print("Grade is: ", grade)
```
